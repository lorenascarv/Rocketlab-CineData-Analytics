```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

spark.sql("CREATE DATABASE IF NOT EXISTS gold")
```

```python
# gold.dim_movies

df_info_silver = spark.table("silver.tb_info_filmes")

df_dim_movies = df_info_silver.withColumn(
    "sk_movie_id", F.row_number().over(Window.orderBy("id_filme"))
)

df_dim_movies_final = df_dim_movies.select(
    "sk_movie_id", "id_filme", "titulo", "data_lancamento", "ano_lancamento",
    "duracao_minutos", "idioma_original", "status_filme", "sinopse"
)

df_dim_movies_final.write.format("delta").mode("overwrite").saveAsTable("gold.dim_movies")
display(df_dim_movies_final)
```

```python
# gold.dim_genres

df_generos_silver = spark.table("silver.tb_generos")

df_dim_genres = (
    df_generos_silver
    .select(F.col("nome_genero").alias("nome_genero"))
    .distinct()
    .withColumn("sk_genre_id", F.row_number().over(Window.orderBy("nome_genero")))
    .select("sk_genre_id", "nome_genero")
)

df_dim_genres.write.format("delta").mode("overwrite").saveAsTable("gold.dim_genres")
display(df_dim_genres)
```

```python
# gold.dim_people

df_pessoas_silver = spark.table("silver.tb_pessoas_empresas")

# apenas pessoas físicas
df_dim_people = (
    df_pessoas_silver
    .filter(F.col("tipo_entidade").isin("Ator", "Diretor", "Roteirista"))
    .select(F.col("nome_pessoa"), F.col("tipo_entidade").alias("tipo_pessoa"))
    .distinct()
    .withColumn("sk_person_id", F.row_number().over(Window.orderBy("nome_pessoa", "tipo_pessoa")))
    .select("sk_person_id", "nome_pessoa", "tipo_pessoa")
)

df_dim_people.write.format("delta").mode("overwrite").saveAsTable("gold.dim_people")
display(df_dim_people)
```

```python
# gold.dim_companies

df_dim_companies = (
    df_pessoas_silver
    .filter(F.col("tipo_entidade") == "Produtora")
    .select(F.col("nome_pessoa").alias("nome_produtora"))
    .distinct()
    .withColumn("sk_company_id", F.row_number().over(Window.orderBy("nome_produtora")))
    .select("sk_company_id", "nome_produtora")
)

df_dim_companies.write.format("delta").mode("overwrite").saveAsTable("gold.dim_companies")
display(df_dim_companies)
```

```python
# gold.dim_reviews

df_reviews_silver = spark.table("silver.tb_avaliacoes_usuarios")

df_dim_reviews = (
    df_reviews_silver
    .groupBy("id_filme")
    .agg(
        F.count("*").alias("qtd_avaliacoes_usuarios"),
        F.round(F.avg("nota_usuario"), 2).alias("nota_media_usuarios")
    )
)

# junta com dim_movies para pegar o sk_movie_id correto
df_dim_reviews = (
    df_dim_reviews
    .join(df_dim_movies_final.select("sk_movie_id", "id_filme"), "id_filme", "inner")
    .withColumn("sk_review_id", F.row_number().over(Window.orderBy("sk_movie_id")))
    .select("sk_review_id", "sk_movie_id", "qtd_avaliacoes_usuarios", "nota_media_usuarios")
)

df_dim_reviews.write.format("delta").mode("overwrite").saveAsTable("gold.dim_reviews")
display(df_dim_reviews)
```

```python
# gold.fact_movies_performance

df_fin_silver = spark.table("silver.tb_financeiro_filmes")
df_metrics_silver = spark.table("silver.tb_metricas_engajamento")

df_fact = (
    df_dim_movies_final.select("sk_movie_id", "id_filme")
    .join(df_fin_silver, "id_filme", "left")
    .join(df_metrics_silver, "id_filme", "left")
)

df_fact_final = df_fact.select(
    "sk_movie_id",
    F.col("orcamento_usd").cast("decimal(18,2)"),
    F.col("receita_usd").cast("decimal(18,2)"),
    F.col("lucro_usd").cast("decimal(18,2)"),
    F.col("orcamento_brl").cast("decimal(18,2)"),
    F.col("receita_brl").cast("decimal(18,2)"),
    F.col("lucro_brl").cast("decimal(18,2)"),
    F.col("popularidade").cast("double"),
    F.col("nota_media_tmdb").cast("double"),
    F.col("qtd_votos_tmdb").cast("int"),
    F.col("nota_media_imdb").cast("double"),
    F.col("qtd_votos_imdb").cast("int"),
)

df_fact_final.write.format("delta").mode("overwrite").saveAsTable("gold.fact_movies_performance")
display(df_fact_final)

# não pode ter sk_movie_id duplicado
print("Linhas totais:", df_fact_final.count())
print("sk_movie_id distintos:", df_fact_final.select("sk_movie_id").distinct().count())
```

```python
# gold.bridge_movie_genre

df_bridge_genre = (
    spark.table("silver.tb_generos")
    .join(spark.table("gold.dim_movies"), "id_filme", "inner")
    .join(spark.table("gold.dim_genres"), "nome_genero", "inner")
    .select("sk_movie_id", "sk_genre_id")
    .distinct()
)
df_bridge_genre.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_genre")
display(df_bridge_genre)

# join pelo par
df_bridge_person = (
    spark.table("silver.tb_pessoas_empresas")
    .filter(F.col("tipo_entidade").isin("Ator", "Diretor", "Roteirista"))
    .withColumnRenamed("tipo_entidade", "tipo_pessoa")
    .join(spark.table("gold.dim_movies"), "id_filme", "inner")
    .join(spark.table("gold.dim_people"), ["nome_pessoa", "tipo_pessoa"], "inner")
    .select("sk_movie_id", "sk_person_id")
    .distinct()
)
df_bridge_person.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_person")
display(df_bridge_person)

df_bridge_company = (
    spark.table("silver.tb_pessoas_empresas")
    .filter(F.col("tipo_entidade") == "Produtora")
    .withColumnRenamed("nome_pessoa", "nome_produtora")
    .join(spark.table("gold.dim_movies"), "id_filme", "inner")
    .join(spark.table("gold.dim_companies"), "nome_produtora", "inner")
    .select("sk_movie_id", "sk_company_id")
    .distinct()
)
df_bridge_company.write.format("delta").mode("overwrite").saveAsTable("gold.bridge_movie_company")
display(df_bridge_company)
```

```python
# gold.gold_genai_movies_context

# diretores e atores por filme
df_atores = (
    spark.table("gold.bridge_movie_person")
    .join(spark.table("gold.dim_people").filter(F.col("tipo_pessoa") == "Ator"), "sk_person_id")
    .groupBy("sk_movie_id")
    .agg(F.concat_ws(", ", F.collect_list("nome_pessoa")).alias("atores_principais"))
)

df_diretores = (
    spark.table("gold.bridge_movie_person")
    .join(spark.table("gold.dim_people").filter(F.col("tipo_pessoa") == "Diretor"), "sk_person_id")
    .groupBy("sk_movie_id")
    .agg(F.concat_ws(", ", F.collect_list("nome_pessoa")).alias("diretores"))
)

# tabela com tratamento
df_context = (
    spark.table("gold.dim_movies")
    .join(spark.table("gold.fact_movies_performance"), "sk_movie_id", "left")
    .join(df_atores, "sk_movie_id", "left")
    .join(df_diretores, "sk_movie_id", "left")
    .select(
        F.col("id_filme").alias("movie_id"),
        F.col("titulo").alias("title"),
        F.concat(
            F.lit("O filme "),
            F.coalesce(F.col("titulo"), F.lit("Sem Título")),
            F.lit(", lançado no ano de "),
            F.coalesce(F.col("ano_lancamento").cast("string"), F.lit("não informado")),
            F.lit(", faturou "),
            F.coalesce(
                F.concat(F.lit("US$ "), F.format_number(F.col("receita_usd"), 2)),
                F.lit("valor não informado")
            ),
            F.lit(" e teve um custo de "),
            F.coalesce(
                F.concat(F.lit("US$ "), F.format_number(F.col("orcamento_usd"), 2)),
                F.lit("valor não informado")
            ),
            F.lit(". Estrelado por "),
            F.coalesce(F.col("atores_principais"), F.lit("elenco não informado")),
            F.lit(" e dirigido por "),
            F.coalesce(F.col("diretores"), F.lit("direção não informada")),
            F.lit(", o filme possui a seguinte sinopse: "),
            F.coalesce(F.col("sinopse"), F.lit("Sem sinopse disponível."))
        ).alias("llm_context_document")
    )
)

df_context.write.format("delta").mode("overwrite").saveAsTable("gold.gold_genai_movies_context")
display(df_context)
```

```python
# Desafio de Analytics

# 1. Qual é a receita total (em R$) somada de todos os filmes da base?
print("--- 1. RECEITA TOTAL (R$) ---")
display(spark.table("gold.fact_movies_performance").select(F.sum("receita_brl").alias("receita_total_brl")))

# 2. Quais são os 5 filmes com maior popularidade? Mostre título e valor de popularidade
print("--- 2. TOP 5 POPULARIDADE ---")
display(
    spark.table("gold.fact_movies_performance")
    .join(spark.table("gold.dim_movies"), "sk_movie_id")
    .select("titulo", "popularidade")
    .orderBy(F.col("popularidade").desc())
    .limit(5)
)

# 3. Quantos filmes cada gênero possui? Liste do maior para o menor volume
print("--- 3. FILMES POR GÊNERO ---")
display(
    spark.table("gold.bridge_movie_genre")
    .join(spark.table("gold.dim_genres"), "sk_genre_id")
    .groupBy("nome_genero")
    .agg(F.countDistinct("sk_movie_id").alias("qtd_filmes"))
    .orderBy(F.col("qtd_filmes").desc())
)

# 4. Para os 10 filmes de maior receita, mostre título, receita (em US$ e R$) e a posição de cada um no ranking (RANK()).
print("--- 4. TOP 10 RECEITA COM RANK ---")
window_rank = Window.orderBy(F.col("receita_usd").desc())
display(
    spark.table("gold.fact_movies_performance")
    .join(spark.table("gold.dim_movies"), "sk_movie_id")
    .withColumn("ranking", F.rank().over(window_rank))
    .select("ranking", "titulo", "receita_usd", "receita_brl")
    .orderBy("ranking")
    .limit(10)
)


max_date = (
    spark.table("gold.dim_movies")
    .filter((F.col("data_lancamento").isNotNull()) & (F.col("data_lancamento") <= F.current_date()))
    .select(F.max(F.col("data_lancamento").cast("date")))
    .collect()[0][0]
)

print(f"Data limite considerada: {max_date}")

# 5. Qual ator teve a maior quantidade de participações nos filmes lançados nos últimos 2 anos*?
print("--- 5. ATOR COM MAIS PARTICIPAÇÕES (ÚLTIMOS 2 ANOS) ---")
display(
    spark.table("gold.dim_movies")
    .withColumn("dt_lanc", F.col("data_lancamento").cast("date"))
    .filter((F.col("dt_lanc").isNotNull()) & (F.col("dt_lanc") >= F.add_months(F.lit(max_date), -24)) & (F.col("dt_lanc") <= F.lit(max_date)))
    .join(spark.table("gold.bridge_movie_person"), "sk_movie_id")
    .join(spark.table("gold.dim_people").filter(F.col("tipo_pessoa") == "Ator"), "sk_person_id")
    .groupBy("nome_pessoa")
    .agg(F.countDistinct("sk_movie_id").alias("qtd_participacoes"))
    .orderBy(F.col("qtd_participacoes").desc())
    .limit(1)
)

# 6. Qual a produtora de filmes teve o maior Lucro nos últimos 5 anos*?
print("--- 6. PRODUTORA COM MAIOR LUCRO (ÚLTIMOS 5 ANOS) ---")
display(
    spark.table("gold.dim_movies")
    .withColumn("dt_lanc", F.col("data_lancamento").cast("date"))
    .filter((F.col("dt_lanc").isNotNull()) & (F.col("dt_lanc") >= F.add_months(F.lit(max_date), -60)) & (F.col("dt_lanc") <= F.lit(max_date)))
    .join(spark.table("gold.fact_movies_performance"), "sk_movie_id")
    .filter(F.col("lucro_usd").isNotNull()) 
    .join(spark.table("gold.bridge_movie_company"), "sk_movie_id")
    .join(spark.table("gold.dim_companies"), "sk_company_id")
    .groupBy("nome_produtora")
    .agg(F.sum("lucro_usd").alias("lucro_total_usd"))
    .orderBy(F.col("lucro_total_usd").desc())
    .limit(1)
)
```
