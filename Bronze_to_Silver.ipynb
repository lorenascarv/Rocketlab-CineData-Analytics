```python
# silver.tb_cotacao_dolar

from pyspark.sql import functions as F
from pyspark.sql.window import Window

df_cotacao_bronze = spark.table("bronze.tb_cotacao_dolar")

df_cotacao_dia = (
    df_cotacao_bronze
    .withColumn("data_cotacao", F.to_date("dataHoraCotacao"))
    .withColumn("cotacao_compra", F.col("cotacaoCompra").cast("decimal(10,4)"))
    .groupBy("data_cotacao")
    .agg(F.max("cotacao_compra").alias("cotacao_compra"))
)

# gera serie temporall contínua
data_min, data_max = df_cotacao_dia.select(F.min("data_cotacao"), F.max("data_cotacao")).first()

df_calendario = spark.sql(f"""
    SELECT explode(sequence(to_date('{data_min}'), to_date('{data_max}'), interval 1 day)) AS data_cotacao
""")

# para cada dia sem cotação, colocar o último dia útil
window_fill = Window.orderBy("data_cotacao").rowsBetween(Window.unboundedPreceding, 0)

df_cotacao_silver = (
    df_calendario
    .join(df_cotacao_dia, "data_cotacao", "left")
    .withColumn("cotacao_compra", F.last("cotacao_compra", ignorenulls=True).over(window_fill))
)

df_cotacao_silver.write.format("delta").mode("overwrite").saveAsTable("silver.tb_cotacao_dolar")
display(df_cotacao_silver)
```

```python
# silver.tb_info_filmes

df_info_bronze = spark.table("bronze.tb_movies_info")

# tabelas para português
df_info = (
    df_info_bronze
    .withColumnRenamed("id", "id_filme")
    .withColumnRenamed("title", "titulo")
    .withColumnRenamed("original_title", "titulo_original")
    .withColumnRenamed("release_date", "data_lancamento_raw")
    .withColumnRenamed("runtime", "duracao_minutos")
    .withColumnRenamed("original_language", "idioma_original")
    .withColumnRenamed("status", "status_raw")
    .withColumnRenamed("overview", "sinopse")
    .withColumnRenamed("tagline", "frase_divulgacao")
)

# normalização da coluna (removendo ruídos, hífens sobressalentes e padronizando a caixa)
df_info = df_info.withColumn(
    "status_normalizado",
    F.trim(F.regexp_replace(F.upper(F.regexp_replace(F.col("status_raw"), r"[-_]+", " ")), r"\s+", " "))
)

# tradução do status
df_info = df_info.withColumn(
    "status_filme",
    F.when(F.col("status_normalizado") == "RELEASED", "Lançado")
     .when(F.col("status_normalizado") == "POST PRODUCTION", "Pós-Produção")
     .when(F.col("status_normalizado") == "IN PRODUCTION", "Em Produção")
     .when(F.col("status_normalizado") == "PLANNED", "Planejado")
     .when(F.col("status_normalizado") == "RUMORED", "Rumores")
     .when(F.col("status_normalizado").isin("CANCELED", "CANCELLED"), "Cancelado")
     .otherwise("Não Informado")
)

# tratamento de Data Multi-Formato
df_info = df_info.withColumn(
    "data_lancamento",
    F.coalesce(
        F.expr("try_to_date(data_lancamento_raw, 'yyyy-MM-dd')"),
        F.expr("try_to_date(data_lancamento_raw, 'MM-dd-yyyy')"),
        F.expr("try_to_date(data_lancamento_raw, 'dd-MM-yyyy')"),
        F.expr("try_to_date(data_lancamento_raw, 'dd/MM/yyyy')"),
        F.expr("try_to_date(data_lancamento_raw, 'MM/dd/yyyy')"),
        F.expr("try_to_date(data_lancamento_raw, 'yyyy/MM/dd')")
    )
)

# coluna derivada
df_info = df_info.withColumn("ano_lancamento", F.year("data_lancamento"))

# deduplicação
window_dedup = Window.partitionBy("id_filme").orderBy(F.col("ingestion_datetime").desc())
df_info = (
    df_info.withColumn("rn", F.row_number().over(window_dedup))
    .filter(F.col("rn") == 1)
    .drop("rn")
)

df_info_final = df_info.select(
    "id_filme", "titulo", "titulo_original", "data_lancamento", "ano_lancamento",
    "duracao_minutos", "idioma_original", "status_filme", "sinopse", "frase_divulgacao"
)

df_info_final.write.format("delta").mode("overwrite").saveAsTable("silver.tb_info_filmes")
display(df_info_final)

total = df_info_final.count()
nulos = df_info_final.filter(F.col("data_lancamento").isNull()).count()
print(f"Total: {total} | Nulos: {nulos} | Percentual: {round(nulos/total*100, 2)}%")
```

```python
# silver.tb_financeiro_filmes
from pyspark.sql import functions as F

df_fin_bronze = spark.table("bronze.tb_movies_financials")
df_cotacao = spark.table("silver.tb_cotacao_dolar")
df_datas = spark.table("silver.tb_info_filmes").select("id_filme", "data_lancamento")

cotacao_fallback = df_cotacao.orderBy(F.col("data_cotacao").desc()).first()["cotacao_compra"]

def limpar_valor_monetario(coluna):
    # ausência de dado para NULL 
    col_tratada = F.when(
        F.trim(F.upper(coluna.cast("string"))).isin("UNKNOWN", "NÃO INFORMADO", "N/A", "NULL", ""), None
    ).otherwise(coluna.cast("string"))
    # remove símbolo de moeda, etc
    col_limpa = F.regexp_replace(col_tratada, r"[^\d,\.\-]", "")
    # remove pontuações de milhar
    col_limpa = F.regexp_replace(col_limpa, r"\.(?=\d{3}(\D|$))", "")
    # troca vírgula decimal por ponto
    col_limpa = F.regexp_replace(col_limpa, r",", ".")
    return col_limpa.cast("decimal(18,2)")

df_fin = (
    df_fin_bronze
    .withColumnRenamed("id", "id_filme")
    .withColumn("orcamento_usd", limpar_valor_monetario(F.col("budget")))
    .withColumn("receita_usd", limpar_valor_monetario(F.col("revenue")))
)

# valores zerados ou negativos
df_fin = (
    df_fin
    .withColumn("orcamento_usd", F.when(F.col("orcamento_usd") <= 0, None).otherwise(F.col("orcamento_usd")))
    .withColumn("receita_usd", F.when(F.col("receita_usd") <= 0, None).otherwise(F.col("receita_usd")))
)

# junta com a data de lançamento pra pegar a cotação do dia
df_fin = (
    df_fin.join(df_datas, "id_filme", "left").alias("f")
    .join(df_cotacao.alias("c"), F.col("f.data_lancamento") == F.col("c.data_cotacao"), "left")
    .select("f.*", F.coalesce(F.col("c.cotacao_compra"), F.lit(cotacao_fallback)).alias("cotacao_compra"))
)

df_fin = df_fin.withColumn("orcamento_brl", F.round(F.col("orcamento_usd") * F.col("cotacao_compra"), 2))
df_fin = df_fin.withColumn("receita_brl", F.round(F.col("receita_usd") * F.col("cotacao_compra"), 2))

# lucro só calcula se orçamento e receita existirem, ambos
df_fin = df_fin.withColumn(
    "lucro_usd",
    F.when(F.col("orcamento_usd").isNotNull() & F.col("receita_usd").isNotNull(),
           F.col("receita_usd") - F.col("orcamento_usd"))
).withColumn(
    "lucro_brl",
    F.when(F.col("orcamento_brl").isNotNull() & F.col("receita_brl").isNotNull(),
           F.col("receita_brl") - F.col("orcamento_brl"))
)

# evita divisão por 0 
df_fin = df_fin.withColumn(
    "margem_lucro_pct",
    F.when(
        F.col("orcamento_usd").isNotNull() & (F.col("orcamento_usd") != 0) & F.col("lucro_usd").isNotNull(),
        F.round((F.col("lucro_usd") / F.col("orcamento_usd")) * 100, 2)
    )
)

from pyspark.sql.window import Window

window_dedup_fin = Window.partitionBy("id_filme").orderBy(F.col("ingestion_datetime").desc())
df_fin = (
    df_fin.withColumn("rn", F.row_number().over(window_dedup_fin))
    .filter(F.col("rn") == 1)
    .drop("rn")
)

df_fin_final = df_fin.select(
    "id_filme", "orcamento_usd", "receita_usd", "orcamento_brl", "receita_brl",
    "lucro_usd", "lucro_brl", "margem_lucro_pct"
)

df_fin_final.write.format("delta").mode("overwrite").saveAsTable("silver.tb_financeiro_filmes")
display(df_fin_final)
```

```python
# silver.tb_metricas_engajamento

df_metrics_bronze = spark.table("bronze.tb_movies_metrics")

# "coluna de popularidade apresenta inconsistências de pontuação e separadores decimais na origem"
df_metrics = df_metrics_bronze.withColumn(
    "_tem_virgula", F.col("popularity").cast("string").contains(",")
)
df_metrics = df_metrics.withColumn(
    "popularidade_limpa",
    F.when(
        F.col("_tem_virgula"),
        F.regexp_replace(F.regexp_replace(F.col("popularity").cast("string"), r"\.", ""), ",", ".")
    ).otherwise(F.col("popularity").cast("string"))
)

# conversão de tipagem segura 
df_metrics = (
    df_metrics
    .withColumnRenamed("id", "id_filme")
    .withColumn("popularidade", F.expr("try_cast(popularidade_limpa AS DOUBLE)"))
    .withColumn("nota_media_tmdb", F.expr("try_cast(vote_average AS DOUBLE)"))
    .withColumn("qtd_votos_tmdb", F.expr("try_cast(vote_count AS INT)"))
    .withColumn("nota_media_imdb", F.expr("try_cast(averageRating AS DOUBLE)"))
    .withColumn("qtd_votos_imdb", F.expr("try_cast(numVotes AS INT)"))
)

# notas fora de 0-10 e contagens/popularidade negativas viram NULL
def validar_nota(c):
    return F.when((c >= 0) & (c <= 10), c)

def validar_nao_negativo(c):
    return F.when(c >= 0, c)

df_metrics = (
    df_metrics
    .withColumn("nota_media_tmdb", validar_nota(F.col("nota_media_tmdb")))
    .withColumn("nota_media_imdb", validar_nota(F.col("nota_media_imdb")))
    .withColumn("popularidade", validar_nao_negativo(F.col("popularidade")))
    .withColumn("qtd_votos_tmdb", validar_nao_negativo(F.col("qtd_votos_tmdb")))
    .withColumn("qtd_votos_imdb", validar_nao_negativo(F.col("qtd_votos_imdb")))
)

window_dedup_metrics = Window.partitionBy("id_filme").orderBy(F.col("ingestion_datetime").desc())
df_metrics = (
    df_metrics.withColumn("rn", F.row_number().over(window_dedup_metrics))
    .filter(F.col("rn") == 1)
    .drop("rn")
)

df_metricas_final = df_metrics.select(
    "id_filme", "popularidade", "nota_media_tmdb", "qtd_votos_tmdb", "nota_media_imdb", "qtd_votos_imdb"
)

df_metricas_final.write.format("delta").mode("overwrite").saveAsTable("silver.tb_metricas_engajamento")
display(df_metricas_final)
```

```python
# silver.tb_avaliacoes_usuarios

df_reviews_bronze = spark.table("bronze.tb_movies_reviews")

df_reviews = (
    df_reviews_bronze
    .withColumnRenamed("id", "id_filme")
    .withColumnRenamed("nome", "nome_usuario")
    .withColumnRenamed("nota", "nota_usuario")
    .withColumnRenamed("comentario", "comentario_usuario")
)

# remove registros integralmente duplicados
df_reviews = df_reviews.dropDuplicates(["id_filme", "nome_usuario", "nota_usuario", "comentario_usuario"])

# nota 0-10, se não vira null
df_reviews = df_reviews.withColumn(
    "nota_usuario",
    F.when((F.col("nota_usuario") >= 0) & (F.col("nota_usuario") <= 10), F.col("nota_usuario"))
)

# comentário vazio ou só espaços em branco recebe "Sem comentário"
df_reviews = df_reviews.withColumn(
    "comentario_usuario",
    F.when(
        F.col("comentario_usuario").isNull() | (F.trim(F.col("comentario_usuario")) == ""),
        F.lit("Sem comentário")
    ).otherwise(F.col("comentario_usuario"))
)

df_reviews_final = df_reviews.select("id_filme", "nome_usuario", "nota_usuario", "comentario_usuario")

df_reviews_final.write.format("delta").mode("overwrite").saveAsTable("silver.tb_avaliacoes_usuarios")
display(df_reviews_final)
```

```python
# silver.tb_generos

df_credits_bronze = spark.table("bronze.tb_credits_and_tags")

# padronizando separadores e Explodir (split + explode) a coluna genres
df_generos = df_credits_bronze.withColumn(
    "genres_padronizado", F.regexp_replace(F.col("genres"), ";", ",")
)
df_generos = df_generos.withColumn(
    "nome_genero", F.explode(F.split(F.col("genres_padronizado"), ","))
)
df_generos = df_generos.withColumn(
    "nome_genero", F.trim(F.col("nome_genero"))
)

# alterei essa parte do código, porque tem muitas informações que não são gêneros de filmes
generos_validos = [
    "Action", "Adventure", "Animation", "Comedy", "Crime", "Documentary",
    "Drama", "Family", "Fantasy", "History", "Horror", "Music", "Mystery",
    "Romance", "Science Fiction", "TV Movie", "Thriller", "War", "Western"
]

df_generos = df_generos.withColumn(
    "nome_genero_normalizado", F.initcap(F.trim(F.col("nome_genero")))
)

df_generos = df_generos.filter(
    F.col("nome_genero_normalizado").isin(generos_validos)
)

df_generos = df_generos.withColumn("nome_genero", F.col("nome_genero_normalizado"))

df_generos_final = (
    df_generos.withColumnRenamed("id", "id_filme")
    .select("id_filme", "nome_genero")
    .dropDuplicates()
)

df_generos_final.write.format("delta").mode("overwrite").saveAsTable("silver.tb_generos")
display(df_generos_final)
```

```python
# silver.tb_pessoas_empresas

# tem muitos dados que não são nomes de pessoas, nome de arquivos, sinopses, etc...
def montar_entidades(coluna_origem, tipo_entidade):
    # única forma que achei de tirar as sinopses
    palavras_sinopse = r"(?i)\b(about|movie|story|styled|land|life|death|long|time|past|aversion|mixture|tradition|with|from|their|there|they|which|where|when|this|that)\b"

    return (
        df_credits_bronze
        .withColumnRenamed("id", "id_filme")
        .withColumn("col_limpa", F.regexp_replace(F.col(coluna_origem), ";", ","))
        .withColumn("col_limpa", F.regexp_replace(F.col("col_limpa"), r'[\"\[\]\\\'`]', ''))
        .withColumn("nome_pessoa", F.explode(F.split(F.col("col_limpa"), ",")))
        .withColumn("nome_pessoa", F.trim(F.col("nome_pessoa")))
        .filter((F.col("nome_pessoa") != "") & (F.length(F.col("nome_pessoa")) <= 40))
        .filter(~F.col("nome_pessoa").rlike(r'(?i)\.(jpg|jpeg|png|gif|webp)$'))
        .filter(~F.col("nome_pessoa").contains("/"))
        .filter(~F.col("nome_pessoa").rlike(r'^[@()0-9>"`\\-]'))
        .filter(~F.col("nome_pessoa").rlike(r'[\.\?\!\_]'))
        .filter(F.size(F.split(F.col("nome_pessoa"), r'\s+')) <= 4)
        .filter(~F.col("nome_pessoa").rlike(palavras_sinopse))
        
        .withColumn("tipo_entidade", F.lit(tipo_entidade))
        .select("id_filme", "nome_pessoa", "tipo_entidade")
    )

df_pessoas_empresas = (
montar_entidades("cast", "Ator")
.union(montar_entidades("directors", "Diretor"))
.union(montar_entidades("writers", "Roteirista"))
.union(montar_entidades("production_companies", "Produtora"))
)

# padroniza titulo e remove duplicatas
df_pessoas_empresas = (
    df_pessoas_empresas
    .withColumn("nome_pessoa", F.initcap(F.trim(F.col("nome_pessoa"))))
    .dropDuplicates(["id_filme", "nome_pessoa", "tipo_entidade"])
)

df_pessoas_empresas.write.format("delta").mode("overwrite").saveAsTable("silver.tb_pessoas_empresas")
display(df_pessoas_empresas)
    
```
